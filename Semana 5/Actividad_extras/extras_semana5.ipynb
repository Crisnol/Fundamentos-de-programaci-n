**Extra 1: Sistema de calificaciones con tuplas**

Enunciado: Crea una tupla calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0). Imprime la tercera calificación, captura dos calificaciones nuevas con input() y anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala de mayor a menor y muestra la suma de todos los elementos usando una función que retorne el valor. Entrada:

Nueva calificación 1: 8.5
Nueva calificación 2: 9.5

Explicación: 

En este ejercicio trabajamos con tuplas y listas. Primero creamos una tupla con 5 calificaciones y sacamos la tercera (índice 2). Luego pedimos 2 nuevas calificaciones al usuario y las agregamos a la tupla original creando una tupla nueva. Después convertimos esa tupla en lista para poder ordenarla de mayor a menor usando .sort(reverse=True). Por ultimo, hacemos una funcion que suma todos los valores y retorna el resultado. 



```python
calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

def sumaCal (list_lv_calificaciones):
    suma = 0
    for valor in list_lv_calificaciones:
        suma += valor
    
    return suma

print(f"Tercera calificación: {calificaciones[2]}")

nCal1 = float(input("Ingresa la nueva calificación: "))
nCal2 = float(input("Ingresa la nueva calificación: "))

lv_calificaciones = calificaciones + (nCal1,nCal2)
list_lv_calificaciones = list(lv_calificaciones)
list_lv_calificaciones.sort(reverse=True)

print(f"Nuevas Calificaciónes: {lv_calificaciones}")
print(f"Calificaciones ordenadas {list_lv_calificaciones}")

sumaCal = sumaCal(list_lv_calificaciones)
print(f"Suma Calificiones {sumaCal}")
```

**Extra 2: Agenda de contactos con búsqueda**

Enunciado: Crea un diccionario agenda con tres contactos iniciales ("Ana": "555-0101", "Luis": "555-0102", "Mía": "555-0103"). Captura un nuevo contacto (nombre y teléfono) con input() y agrégalo al diccionario. Luego imprime todos los nombres con keys() en una sola línea y, con una función que retorne el valor, busca y muestra el teléfono del nuevo contacto. Entrada:

Nombre del nuevo contacto: Pedro
Teléfono del nuevo contacto: 555-0140
Nombre a buscar: Pedro

Explicación: 

Aqui usamos diccionarios para guardar contactos. Empezamos con 3 contactos iniciales (nombre: teléfono). El usuario agrega un nuevo contacto con input(). Despues imprimimos todos los nombres usando .keys() en un bucle. Creamos una función que busca el teléfono de un contacto usando .get(), que es mas seguro porque si no encuentra el contacto devuelve un mensaje en lugar de error

```python
def buscarNumero(nombre):
    telefono = contactos.get(nombre, "Contacto no encontrado")
    print(f"\nEl teléfono {nombre} es: {telefono}")


contactos = {
    "ana": "555-0101", 
    "luis": "555-0102", 
    "mia": "555-0103"
}

nombre = input("Ingresa el nombre del nuevo contacto: ")
numero = input("Ingresa su numero")

contactos[nombre] = numero

print("Contactos Registrados: ", end="")
for nombre in contactos.keys():
    print(nombre, end=", ")
    
nombre = input("Ingresa el nombre a buscar: ")
buscarNumero(nombre.lower())
```

**Extra 3: Calculadora segura con manejo de excepciones**

Enunciado: Escribe un programa que pida dos números enteros. Implementa un bloque try-except que capture ValueError (caracteres no numéricos o vacíos) con un mensaje controlado y un bloque específico except ZeroDivisionError que muestre un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la división del primero entre el segundo. Entrada:

Primer número entero: 20
Segundo número entero: 4

Explicación: 

Pedimos dos números enteros y los metemos en un try-except. Si el usuario pone texto o deja vacio, catch el ValueError y pedimos de nuevo con un while. Luego intentamos dividir en otro bloque try-except separado para el ZeroDivisionError (cuando divides entre cero)

```python
while True:
    try:
        num1 = int(input("Ingrese el primer numero entero"))
        num2 = int(input("Ingresa el segundo numero entero"))
        try:
            divicion = num1 / num2
            print(f"División entre {num1} / {num2} = {divicion}")
            break
        except ZeroDivisionError:
            print("Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
    except:
        print("Error: El número es incorrecto")
        
```

**Extra 4: Analizador de mensajes con strings**

Enunciado: Crea una variable mensaje = "Python es un lenguaje poderoso". Imprime su longitud con len(), conviértelo a mayúsculas, reemplaza la palabra "Python" por "programación" y, con una función que retorne el conteo, muestra cuántas palabras contiene el mensaje original. Entrada:

mensaje = "Python es un lenguaje poderoso"

Explicación: 

Tomamos un mensaje y hacemos varias cosas. Primero sacamos su longitud con len(). Despues lo convertimos a mayúsculas con .upper(). Reemplazamos la palabra "Python" por "programación" con .replace(). Y hacemos una función que cuenta cuántas palabras hay dividiendo el mensaje con .split() y contando los elementos.

```python
mensaje = "Python es un lenguaje poderoso"
def conteoPalbras(mensaje):
    palabras = mensaje.split()
    total_palabras = len(palabras)
    return total_palabras

count_frase = len(mensaje)
mayus_frase = mensaje.upper()
replace_frase = mensaje.replace("Python", "programación")

total_palabras = conteoPalbras(mensaje)
print(f"Frase Original: {mensaje} - Longitud: {count_frase} - Cantidad Palabra: {total_palabras}")
print(f"Frase UPPER: {mayus_frase}")
print(f"Frase con Remplazo: {replace_frase}")


```

**Desafío Extra 2: Contador de votos con diccionario**

Crea un diccionario votos = {"Rojo": 0, "Azul": 0, "Verde": 0}. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con items() el color ganador junto con su total.


Explicación:

Creamos un diccionario con 3 candidatos en cero votos. El usuario vota varias veces y sumamos 1 al candidato que votó. Validamos que el voto sea un candidato válido usando .get(). Al final imprimimos todos los resultados con .items() para ver nombre y cantidad de votos.


```python
canditados = {
    "A": 0,
    "B": 0,
    "C": 0
}

cantVotos = int(input("Cantidasd de votos? "))

i = 1
while i <= cantVotos:
    voto = input("Ingresa tu voto: ").upper()
    
    if canditados.get(voto, "Voto incorrecto") == "Voto incorrecto":
        print("El candidato no existe, ingresa uno correcto")
    else:
        print("Voto realizado correctamente")
        canditados[voto.upper()] += 1
        i += 1
    

for k, v in canditados.items():
    print(f"{k} - votos: {v}")
```
