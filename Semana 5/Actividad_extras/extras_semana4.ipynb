**Desafío Extra 1: Lista de espera con tuplas**

Crea una tupla espera = ("María", "José", "Carlos", "Lucía", "Pedro"). Imprime el tercer elemento, captura dos nombres nuevos con input(), anéxalos en una nueva tupla, conviértela en lista, ordénala alfabéticamente y muestra la cantidad total de personas.

```python
espera = ("María", "José", "Carlos", "Lucía", "Pedro")

print(f"3 persona: {espera[2]}")

nuevo1 = input("Ingresa el nombre a ingresar: ")
nuevo2 = input("Ingresa el nombre a ingresar: ")

espera_extendida = espera + (nuevo1, nuevo2)

espera_lista = list(espera_extendida)
espera_lista.sort()
cant_personas = len(espera_lista)

print(f"{espera_lista} - Cantidad Personas: {cant_personas}")
```

**Desafío Extra 2: Contador de votos con diccionario**

Crea un diccionario votos = {"Rojo": 0, "Azul": 0, "Verde": 0}. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con items() el color ganador junto con su total.

D

```python
canditados = {
    "A": 0,
    "B": 0,
    "C": 0
}

cantVotos = int(input("Cantidasd de votos? "))

i = 1
while i <= cantVotos:
    voto = input("Ingresa tu voto: ").upper()
    
    if canditados.get(voto, "Voto incorrecto") == "Voto incorrecto":
        print("El candidato no existe, ingresa uno correcto")
    else:
        print("Voto realizado correctamente")
        canditados[voto.upper()] += 1
        i += 1
    

for k, v in canditados.items():
    print(f"{k} - votos: {v}")
```

**Desafío Extra 3: Validación robusta de entrada**

Escribe un programa que pida al usuario dos números enteros dentro de un bloque try-except. Si el usuario ingresa texto o un valor vacío, muestra un mensaje controlado y repite la solicitud con un ciclo while. Luego pide la división del primero entre el segundo, con un bloque específico para ZeroDivisionError.

```python
while True:
    try:
        num1 = int(input("Ingrese el primer numero"))
        num2 = int(input("Ingresa el segundo numero"))
        break 
    except:
        print("Número Incorrecto")\
            
try:
    divicion = num1 / num2
    print(f"La divición es: {divicion}")
except ZeroDivisionError:
    print("No se puede realizar la división")
```

**Desafío Extra 4: Analizador de textos**

Crea un programa que reciba una frase, imprima su longitud con len(), la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.

```python
frase = input("Escribe una frase: ")

def conteo_frase(frase):
    return len(frase)
    

count_frase = len(frase)
mayus_frase = frase.upper()
replace_frase = frase.replace("Python", "programación")

count_frase_replace = conteo_frase(replace_frase)
print(f"Frase Original: {frase} - Longitud: {count_frase}")
print(f"Frase UPPER: {mayus_frase}")
print(f"Frase con Remplazo: {replace_frase} - Longitud: {count_frase_replace}")
```

**Desafío Extra 5: Minimenú modular integrador**

Construye un menú con un ciclo while y opciones numéricas que permita ejecutar de forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en el diccionario contactos, (3) dividir dos números con manejo de ZeroDivisionError y (4) analizar un mensaje de texto. La opción (5) debe finalizar el programa.

```python
def mostrarTuplaOrdenada (tupla):
    espera_lista = list(tupla)
    espera_lista.sort()
    print(f"Lista ordenada: {espera_lista}")
    print("----------------------------------------\n")

def buscarTelefono(nombre):
    contactos = {
    "juan": "342 2344 2344",
    "nose": "943 2345 9822",
    "cr7": "242 8092 8493"
    }
    telefono = contactos.get(nombre, "Contacto no encontrado")

    print(f"El numero de {nombre} es: {telefono}")
    print("----------------------------------------\n")

def dividirNumero():
    while True:
        try:
            num1 = int(input("Ingrese el primer numero"))
            num2 = int(input("Ingresa el segundo numero"))
            break 
        except:
            print("Número Incorrecto")    
    try:
        divicion = num1 / num2
        print(f"La divición es: {divicion}")
        print("----------------------------------------\n")

    except ZeroDivisionError:
        print("No se puede realizar la división")
        print("----------------------------------------\n")

        
        
def analizarMensaje(frase):
    count_frase = len(frase)
    mayus_frase = frase.upper()
    replace_frase = frase.replace("hola", "(Palabra encriptada)")

    count_frase_replace = len(replace_frase)
    print(f"\nFrase Original: {frase} - Longitud: {count_frase}")
    print(f"Frase UPPER: {mayus_frase}")
    print(f"Frase con Remplazo: {replace_frase} - Longitud: {count_frase_replace}")
    print("----------------------------------------\n")
  
while True:
    print("________________________________________\n")
    print("1) Mostrar tupla ordenada")
    print("2) Buscar telefono")
    print("3) Dividir dos Números")
    print("4) Analizar mensaje")
    print("5) Finalizar el programa")
    print("________________________________________\n")

    
    opcion = int(input("Opción: "))
    
    match opcion:
        case 1:
            print("=========Mostrar tupla ordenada=======\n")
            tupla = (34854, 7684, 34563456, 6556, 7)
            mostrarTuplaOrdenada(tupla)
        case 2:
            print("=========Buscar telefono=======\n")
            nombre = input("Nombre: ")
            buscarTelefono(nombre.lower())
        case 3:
            print("=========Dividir dos Números=======\n")
            dividirNumero()
        case 4:
            print("=========Analizar Mensaje=======\n")
            frase = input("Escribe una frase: ")
            analizarMensaje(frase)
        case 5:
            print("------------------Finalizando Programa, Gracias por su preferencia--------------------------")
            break
        case _:
            print("Opción no valida---------------------\n")
```
