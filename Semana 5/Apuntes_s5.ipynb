## TUPLAS

Una tupla es una secuencia ordenada e inmutable de elementos en Python. A diferencia de las listas, una vez creada una tupla no se pueden agregar, eliminar ni modificar sus elementos. Se define con paréntesis ().

```python
numeros = (19,81,43,53)

frutas = ("manzana", "hola", "noa")

datosMix = ("Nose", 23, 232.23, True)

print(type(numeros))
print(type(frutas))
print(type(datosMix))

lista = [5,3]
tupla_1_elemento = (5,)

print(type(lista))
print(type(tupla_1_elemento))
```

```python
numeros = (19, 81, 43, 53, 89)

print(numeros[0])
print(numeros[2])
print(numeros[-1])
 
```

```python
cordenadas = (5,2)

x, y = cordenadas

print(f"Coordenada X: {x}")
print(f"Coordenada Y: {y}")
```

```python
calificaciones = (23, 64, 98, 78, 90, 0)

cal_extra1 = int(input("Calificacion Nueva"))
cal_extra2 = int(input("Calificacion Nueva"))

nuevasCal = calificaciones + (cal_extra1,cal_extra2)

print(calificaciones)
print(nuevasCal)
```

```python
tupla_desordenada = (9, 3, 2, 9, 12, 89, 2342323423, -4234823)
print(tupla_desordenada)
lista = list(tupla_desordenada)
print()
print(type(lista))
print(lista)

print()
lista.sort()
print(lista)

ordenada = tuple(lista)
print()
print(type(ordenada))
print(ordenada)
```

```python
def suma_tupla (t):
    total = 0
    for e in t:
        total += e
    return total

calorias = (800, 1900, 456, 354, 123)

resultado = suma_tupla(calorias)

print(f"Resultado Calorias: {resultado}")
```

## TEMA 14: DICCIONARIOS

Un diccionario es una colección de pares clave : valor encerrados entre llaves {}. Permite guardar datos asociados, por ejemplo, el nombre de un contacto (clave) con su número de teléfono (valor).

```python
contactos = {
    "Juan": "342 2344 2344",
    "nose": "943 2345 9822",
    "cr7": "242 8092 8493"
}

print(contactos["cr7"])
print(contactos.get("messi","sin registro"))

contactos["Messi"] = "098490234234"
print(contactos)
```

```python
print(contactos.keys())
print(contactos.values())
print(contactos.items())
```

```python
print("Impresicon keys (nombres contactos)")
for n in contactos.keys():
    print(n)

print("\nImpresion values (numeros)")
for v in contactos.values():
    print(v)

print("\nImprecion y asignacion de cada item")
for n,t in contactos.items():
    print(f"Nombre: {n} - Num: {t}")
```

```python
def buscarNumero(diccionario,nombre):
    resultado = diccionario.get(nombre, "Contacto no encontrado")
    return resultado

nombre = input("Nombre: ")
telefono = buscarNumero(contactos,nombre)

print(f"El numero de {nombre} es: {telefono}")
```

## TEMA 15: EXCEPCIONES

Una excepción es un error que ocurre durante la ejecución del programa y que, si no se maneja, provoca que el programa se detenga. Con try/except podemos anticipar estos errores y desplegar mensajes controlados en lugar de crashear.

```python
try:
    edad = int(input("Ingresa tu edad: "))
    print("Edad registrada:", edad)

except ValueError:
    print("El dato ingresado no es un número entero.x")
```

```python
try:
    dividendo = int(input("dividendo"))
    divisor = int(input("divisor"))
    resultado = dividendo/divisor

except ZeroDivisionError:
    print("no se puede dividir entre 0")
else:
    print("nose que paso")
finally:
    print("es finally saludos")
```

```python
contactos = {
    "juan": "342 2344 2344",
    "nose": "943 2345 9822",
    "cr7": "242 8092 8493"
}

def buscarNumero(diccionario,nombre):
    resultado = diccionario.get(nombre, "Contacto no encontrado")
    return resultado

nombre = input("Nombre: ")
telefono = buscarNumero(contactos,nombre.lower())

print(f"El numero de {nombre} es: {telefono}")
```

**Desafío Extra 2: Contador de votos con diccionario**

Crea un diccionario votos = {"Rojo": 0, "Azul": 0, "Verde": 0}. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con items() el color ganador junto con su total.

```python
canditados = {
    "A": 0,
    "B": 0,
    "C": 0
}

cantVotos = int(input("Cantidasd de votos? "))

i = 1
while i <= cantVotos:
    voto = input("Ingresa tu voto: ").upper()
    
    if canditados.get(voto, "Voto incorrecto") == "Voto incorrecto":
        print("El candidato no existe, ingresa uno correcto")
    else:
        print("Voto realizado correctamente")
        canditados[voto.upper()] += 1
        i += 1
    

for k, v in canditados.items():
    print(f"{k} - votos: {v}")
    
    
    
    
```

**Desafío Extra 4: Analizador de textos**

Crea un programa que reciba una frase, imprima su longitud con len(), la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo

```python
frase = input("Escribe una frase: ")

def conteo_frase(frase):
    return len(frase)
    

count_frase = len(frase)
mayus_frase = frase.upper()
replace_frase = frase.replace("Python", "programación")

count_frase_replace = conteo_frase(replace_frase)
print(f"Frase Original: {frase} - Longitud: {count_frase}")
print(f"Frase UPPER: {mayus_frase}")
print(f"Frase con Remplazo: {replace_frase} - Longitud: {count_frase_replace}")
```

**Desafío Extra 5: Minimenú modular integrador**

Construye un menú con un ciclo while y opciones numéricas que permita ejecutar de forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en el diccionario contactos, (3) dividir dos números con manejo de ZeroDivisionError y (4) analizar un mensaje de texto. La opción (5) debe finalizar el programa.

```python
while True:
    print("1) Mostrar tupla ordenada")
    print("2) Buscar telefono")
    print("3) Dividir dos Números")
    print("4) Analizar mensaje")
    print("5) Finalizar el programa")
    
    opcion = int(input("Opción: "))
    
    match opcion:
        case 1:
            print()
        case 2:
            print()
        case 3:
            print()
        case 4:
            print()
        case 5:
            print()
```
