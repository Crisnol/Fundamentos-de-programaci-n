### Extra 1: División de cuenta con propina
Primero se generan las peticiones(total de cuenta, propina y cantidad de personas) para el usuario, una vez hecho esto se guardan en varibles. Despues de esto se hace el procedimiento con las debidas operaciones para conocer nuevos valores como el monto de la propina, etc. y para finalizar se imprimen los valores obtenido que seran de utilidad para el usuario.

```python
total_cuenta = float(input("Ingrese el Total de la cuenta a pagar: "))
porcentaje_propina = int(input("Ingresa el porcentaje de propina: "))
num_personas = int(input("Ingresa el numero de personas: "))

monto_propina = (total_cuenta*porcentaje_propina)/100
total = total_cuenta + monto_propina
porPersona = total/num_personas

print(f"Propina: ${monto_propina:.2f}")
print(f"Total con propina: ${total:.2f}")
print(f"Pago por persona: ${porPersona:.2f}")
```

### Extra 2: Conversor de minutos a días, horas y minutos
Aqui el procedimiento fue primero pedir el valor necesario en este caso lo minutos pidiendolo como int, despues se realiza el procedimieto que son los calculos para determinar cuantos dias, horas y minutos hay para eso se ocupan las conversion de tiempo junto con las operaciones de division entera y el modulo, ya par finalizar se imprimen los datos de manera estructurada. 

```python
input_minutos = int(input("Ingrese una cantidad de minutos (entero): "))

dias = input_minutos//1440
residuo_dias = (input_minutos%1440)
horas = residuo_dias//60
minutos = residuo_dias%60

print(f"{input_minutos} minutos = {dias} día(s), {horas} hora(s), {minutos} minuto(s)")
```

### Extra 3: Calificación final ponderada
En este se pide las 3 calificaciones y se gaurda como float, con eso se hace el procedimiento para obtener, para eso la cal1 y cal2 si multiplica cada por 0.3 y para la cal3 se multiplica por 0.4, los resultados se suman y con eso ya tenemos la cal fina, por ulimo se imprime para que el usuario conosca su calificacion final.## Extra 3: Calificación final ponderada

```python
cal1 = float(input("Ingresa la calificacion del Paricial 1 (30%): "))
cal2 = float(input("Ingresa la calificacion del Paricial 2 (30%): "))
cal3 = float(input("Ingresa la calificacion del Paricial 3 (40%): "))

calificacion_final = (cal1 * 0.30) + (cal2 * 0.30) + (cal3 * 0.40)

print(f"Tu calificacion fian es: {calificacion_final:.2f}")
```

### Extra 4: Conversor de moneda (MXN a USD y EUR)
Para este los valores de entrada se guardad en float ya que manejamos decimales una vez de pedir los tipo de cambio y la cantidad de pesos mexicanos, una vez obtenido eso hago el procedimiento que son las conversiones, para finalizar se imprimio los resultados obtenidos con el formato correspondiente

```python
cant_mxn = float(input("Ingrese la cantidad de peso mexicano (MXN): "))
dolar = float(input("Ingrese el tipo de cambio del dolar (USD): "))
euro = float(input("Ingrese el tipo de cambio del euro (EUR): "))

mxn_dolar = cant_mxn/dolar
mxn_euro = cant_mxn/euro

print(f"${cant_mxn:.2f} MXN equivalen a:")
print(f"USD: {mxn_dolar:.2f}")
print(f"EUR: {mxn_euro:.2f}")
```
