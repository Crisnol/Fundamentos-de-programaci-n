Dada la matriz [[3, 1, 4], [1, 5, 9], [2, 6, 5]], imprime la matriz y la suma de cada fila. Entrada:

Matriz: [[3, 1, 4], [1, 5, 9], [2, 6, 5]]

```python
matriz = [[3, 0, 0], [9, 99, 9], [2, 2, 98]]

def imprisión_matriz(matriz):
    for l in matriz:
        for c in l:
            print(c, end=" ")
        print("")
        
def suma_fila(fila):
    suma = 0
    for e in fila:
        suma = suma + e
    
    return suma

imprisión_matriz(matriz)
for l in range(len(matriz)):
    value = suma_fila(matriz[l])
    print(f"Suma de la fila {l}: {value}")
```
