**Extra 1: Suma de todos los elementos de una matriz**

Enunciado: Crea una función suma_matriz(matriz) que regrese valor (con return) la suma de todos los elementos de una lista de listas, usando ciclos for anidados. Define además una función imprimir_matriz(matriz) que no regrese valor y muestre la matriz sin corchetes ni comas. Aplica ambas funciones a la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9]]. Entrada:

Matriz: [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

Procedimiento: 

Se crea una función para recorrer todas las filas y columnas de la matriz usando ciclos anidados, cada numero se va acumulando en una variable llamada suma, tambien se hace otra función para imprimir la matriz sin corchetes ni comas y al final se muestra la suma total.


```python
matriz = [[30, 7, 6], [8, 4, 6], [1, 9, 3]]

def suma_matriz(matriz):
    suma = 0
    for l in matriz:
        for c in l:
            suma = suma + c
    
    return suma

def imprisión_matriz(matriz):
    
    for l in matriz:
        for c in l:
            print(c, end=" ")
        print("")
            
imprisión_matriz(matriz)
sumaTotal = suma_matriz(matriz)

print(f"La suma de los elementos es: {sumaTotal}")
```

**Extra 2: Matriz con suma por fila**

Enunciado: Crea una función suma_fila(fila) que regrese valor (con return) la suma de los elementos de una lista, y una función mostrar_tabla(matriz) que no regrese valor e imprima la matriz sin corchetes ni comas. Dada la matriz [[3, 1, 4], [1, 5, 9], [2, 6, 5]], imprime la matriz y la suma de cada fila. Entrada:

Matriz: [[3, 1, 4], [1, 5, 9], [2, 6, 5]]

Procedimiento: 

La función suma_fila recibe una fila y suma todos sus elementos, despues se recorre la matriz y se manda cada fila a la función, de esta manera se obtiene y se imprime la suma de cada renglon.

```python
matriz = [[3, 0, 0], [9, 99, 9], [2, 2, 98]]

def imprisión_matriz(matriz):
    for l in matriz:
        for c in l:
            print(c, end=" ")
        print("")
        
def suma_fila(fila):
    suma = 0
    for e in fila:
        suma = suma + e
    
    return suma

imprisión_matriz(matriz)
for l in range(len(matriz)):
    value = suma_fila(matriz[l])
    print(f"Suma de la fila {l}: {value}")
```

**Extra 3: Producto de una columna sin usar el operador**

Enunciado: Crea la matriz [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]] (4 filas y 3 columnas). Define una función multiplicar_columna(matriz, columna) que regrese valor (con return) el producto de los elementos de una columna dada. Restricción: dentro de la función está prohibido usar el operador *; el producto debe calcularse con sumas repetidas usando un ciclo. Pide al usuario el número de columna (0, 1 o 2) y muestra el resultado. Entrada:

Numero de columna: 1

Procedimient:

El programa pide el numero de columna que se desea consultar, despues recorre todas las filas y toma el valor que se encuentra en esa columna para calcular el producto, al final se muestra el resultado obtenido de la columna seleccionada.

```python
matriz = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9],
          [2, 4, 6]]

def multiplicar_columna(matriz, columna):
    total = 1
    
    for i in range (len(matriz)):
        total = total * matriz[i][columna]
        
    return total

columna = int(input("Ingresa la columna que quieres: "))

valor = multiplicar_columna(matriz,columna)

print(f"Producto de la comuna {columna}: {valor}")


```

**Extra 5 arbol negativos positivos**

Procedimiento: 

Se capturan diez numeros dentro del rango permitido, incluyendo valores positivos y negativos, despues se ordena la lista para encontrar el numero negativo de mayor tamaño, con espacios, asteriscos y una barra vertical se dibuja una representación de cada numero

```python
lista = []

count = 0

while count < 10:
    # range -10 hasta el 10
    num = int(input("Ingresa un numero del -10 al 10"))
    if (num >= -10 and num <= 10):
        lista.append(num)
        count += 1
    else:
        print("Número fuera del rango, ingresa uno correcto.")


lista_inv = sorted(lista, reverse=True)

mayor_negativo = lista_inv[-1]

print(lista)

for i in range(len(lista)):
    if(lista[i] > 0):
        print(" " * (mayor_negativo * -1), end="")
        print("|", end="")
        print("*" * lista[i], end="\n")
    else:
        espacio_faltante = lista[i] - mayor_negativo
        print(" " * espacio_faltante, end="")
        print("*" * (lista[i] * -1), end= "")
        print("|", end= "\n")
```
