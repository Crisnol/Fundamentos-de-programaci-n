```python
def promedio (total, cant_cal):
    promedio = total / cant_cal
    return promedio

total = 0

cant_cal = int(input("Ingresa la cantidad de numeros"))

for i in range(1, cant_cal + 1):
    cal = float(input("Calificación: "))
    total = cal + total

print(promedio(total,cant_cal))
```

```python
num_matriz = int(input("Ingresa el tamaño de la matriz"))
matriz = []
i = 1
for f in range(num_matriz):
    fila_actual = []
    for c in range(num_matriz):
        
        inv = num_matriz - i    
        if(c == inv):
            fila_actual.append(1)
            i += 1
        elif(c == f):
            fila_actual.append(1)
        else:
            fila_actual.append(0)
    
                
    matriz.append(fila_actual)
    

print("Impresión Matriz")
for f in range(num_matriz):
    for c in range(num_matriz):
        print(matriz[f][c], end= " ")
    print()
```

```python
lista = []

count = 0

while count < 10:
    # range -10 hasta el 10
    num = int(input("Ingresa un numero del -10 al 10"))
    if (num >= -10 and num <= 10):
        lista.append(num)
        count += 1
    else:
        print("Numero fuera del rango, ingresa uno correcto.")
        continue

# for i in range(10):
#     num = int(input("Ingresa un numero del -10 al 10: "))
#     lista.append(num)

lista_inv = sorted(lista, reverse=True)

mayor_negativo = lista_inv[-1]

print(lista)

for i in range(len(lista)):
    if(lista[i] > 0):
        print(" " * (mayor_negativo * -1), end="")
        print("|", end="")
        print("*" * lista[i], end="\n")
    else:
        espacio_faltante = lista[i] - mayor_negativo
        print(" " * espacio_faltante, end="")
        print("*" * (lista[i] * -1), end= "")
        print("|", end= "\n")
   
    
```
