**Ejercicio 10: Suma acumulada con while**

Pide números mientras sean distintos de 0 (el 0 termina el ciclo y no se suma). Entrada: 5, 8, 2, 0 → Salida: La suma de los números es: 15

```python
suma = 0

while True:
    num = int(input("Ingrese un número / ingrese 0 para finalizar: "))
    if num == 0:
        break
    else:
        suma += num
print(f"La suma de los números es: {suma}")
```

**Ejercicio 13: Suma de una lista con función que regresa valor (sin sum())**

Entrada: [10, 20, 30, 40] → Salida: La suma de los elementos es: 100



```python
def calcular_suma (lista):
    suma = 0
    for i in lista:
        suma += i
    return suma
lista =[10, 20, 30, 40]
sumaLista = calcular_suma(lista)
print(f"La suma de los elementos {sumaLista}")
```

**Ejercicio 14: Máximo de una lista con función que regresa valor (sin max())**

Entrada: [7, 3, 9, 2] → Salida: El elemento máximo es: 9

```python
def conocer_max_lista(lista):
    max_valor = lista[0]
    for elemento in lista:
        if elemento > max_valor:
            max_valor = elemento
    return max_valor

lista = [7, 3, 9, 2]
max_elemento = conocer_max_lista(lista)
print(f"El elemento máximo es: {max_elemento}")
```

**Ejercicio 15: Matriz y suma de la diagonal principal**

Entrada: [[1,2,3],[4,5,6],[7,8,9]] → Salida: La suma de la diagonal principal es: 15

```python
matriz =  [[1,2,3],
           [4,5,6],
           [7,8,9]]

suma = 0

for f in  range (len(matriz)):
    for c in range(len(matriz)):
        if c == f:
            suma += matriz[f][c]

print(f"La suma de la diagonal principal es: {suma}")
```

**Ejercicio 18: Factorial con función recursiva**

Entrada: 5 → Salida: El factorial de 5 es: 120

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)

num = int(input("Ingrese el numero: "))
resultado = factorial(num)
print(f"El factorial es: {resultado}") 
```
