```python
año = int(input("Introduce un año: "))

if año < 1582:
	print("No esta dentro")
else:
	if año % 4 != 0:
		print("Año normal")
	elif año % 100 != 0:
		print("Año Bisiesto")
	elif año % 400 != 0:
		print("Año normal")
	else:
		print("Año Bisiesto")
```

```python
bloques = int(input("Ingresa la cantidad de bloques: "))

altura = 0
bloques_necesarios = 1

while bloques >= bloques_necesarios:
    bloques -= bloques_necesarios  
    altura += 1                  
    bloques_necesarios += 1      

print("La altura de la pirámide es:", altura)
```

```python
lista_original = [1, 2, 4, 4, 1, 4, 2, 6, 2, 9]

lista_sin_repetidos = []

for numero in lista_original:
    if numero not in lista_sin_repetidos:
        lista_sin_repetidos.append(numero)

print("Lista original:", lista_original)
print("Lista sin duplicados:", lista_sin_repetidos)
```

```python
def litors_km_millas_galon(litros):
    galon = litros / 3.785411784
    millas = 100 * 1000 / 1609.344
    return millas / galon

def millas_galon_km_litros(millas):
    km = millas * 1609.344 / 1000 / 100
    litros = 3.785411784
    return litros / km

print(litors_km_millas_galon(3.9))
print(litors_km_millas_galon(7.5))
print(litors_km_millas_galon(10.))
print(millas_galon_km_litros(60.3))
print(millas_galon_km_litros(31.4))
print(millas_galon_km_litros(23.5))
```
