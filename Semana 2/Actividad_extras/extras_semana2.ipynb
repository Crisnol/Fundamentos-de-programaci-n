**Extra 1: Control de aforo con break y continue (while)**

Enunciado: Crea un programa que procese visitantes del museo con un ciclo while. Por cada visitante se captura el costo de su boleto ($0 para menores de 3 años, $30 para 3 a 17 años, $45 para mayores de 18). Si el boleto es gratis, muestra "Menor de 3 años, sin cargo" y continúa con el siguiente visitante (usa continue). Si el total acumulado alcanza o supera $100, detén el ciclo con break y muestra el total.

Procedimiento:

Para resolver este problema, primero identifique que el ciclo se termina cuando alcanza 100 o mayor con un break por lo que el ciclo va a tener "True" para que sea infinito hazta que se cumpla el if y termine con el break, despues se ponen los input y otro comprobación para ver si el costo es 0 e imprimir que no hay cargo, esto para implementar el continue.

```python
i = 1
total = 0

while True:
    costo_boleto = int(input(f"Costo del boleto {i}: "))

    if(costo_boleto == 0):
        print("Menor de 3 años, sin cargo")
        i += 1
        continue

    total = total + costo_boleto
    i += 1

    if(total >= 100):
        print("Total alcanzado, se detiene registro")
        break

print(f"Total acumulado: ${total}")
```

**Extra 2: Estadística de visitantes con for**

Enunciado: Crea un programa que pida el número total de visitantes y, con un ciclo for, capture la edad de cada uno. Debe contar cuántos son adultos (18 años o más) y calcular el promedio de edad de todos los visitantes, mostrando ambos resultados con dos decimales.

Procedimiento:

Empece planteando el For y estableciendo su valores como i y su limite (num_visitante), despues define las variables para los datos y acumuladores a guardar, para finalizar estableci las comprobaciones y la impresion de lo solicitado

```python
num_visitantes = int(input("Ingresa el numero de visitantes: "))
contadorAdultos = 0
sumaEdad = 0

for i in range (1, num_visitantes + 1):
    age = int(input(f"Edad del visitante {i}: "))

    if(age >= 18):
        contadorAdultos = contadorAdultos + 1

    sumaEdad = sumaEdad + age

promedioEdad = sumaEdad/num_visitantes

print(f"Adultos: {contadorAdultos}")
print(f"Promedio edad: {promedioEdad:.2f}")
```

**Extra 3: Depuración de un cobro con PDB**

Enunciado: El siguiente código pretende calcular el precio final de un boleto de $45 con el descuento del adulto mayor (12%) pero contiene un error de lógica. Inserta pdb.set_trace() para rastrear las variables, identifica el error y escribe la versión corregida.

Procedimiento:

Para la depuración importe las herramientas necesarias que son pdb y la colocacion de la función de set_trace() la coloque antes del print para analizar las variables, despues corregi el codigo y lo comprobe otra vez.

```python
import pdb

precio = 45
descuento = 0.12
total = precio - (precio * descuento)
pdb.set_trace()
print(f"Total: ${total:.2f}")
```

**Extra 4: Pirámide de asteriscos con for**

Enunciado: Crea un programa que pida la altura de una pirámide y la dibuje con asteriscos usando ciclos for anidados y el parámetro end de print

Procedimiento:

Para la resolucion de este primero comprendi la funcion de un doble for aninado luego hice la estructura y estableci los rangos de cada uno, despues aplico el print con end, probe el codigo y al funcionar corretamente lo finalicé.


```python
altura = int(input("Ingresa el numero de la altura de la piramide en enteros: "))

for i in range(1, altura + 1):
    for j in range(1, i + 1 ):
        print("*", end="")
    print()
```
