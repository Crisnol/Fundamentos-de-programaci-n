**Ejercicio 1: Mayor de edad**

Enunciado: Crea un programa que pida la edad de una persona y muestre si es mayor de edad (18 años o más) o menor de edad. Entrada: 17 Salida: Eres menor de edad

```python
age = int( input("Ingrese su edad:"))

if(age >= 18):
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
```

**Ejercicio 2: Número par o impar**

Enunciado: Crea un programa que pida un número entero y determine si es par o impar usando el operador módulo %. Entrada: 7 Salida: 7 es un número impar

```python
num = int(input("Ingresa un numero entero: "))

if(num % 2 == 0):
    print("Es un numero par")
else:
    print("Es impar")
```

**Ejercicio 3: Calificación aprobatoria**

Enunciado: Crea un programa que pida una calificación de 0 a 10 y muestre si la materia está aprobada (mayor o igual a 6) o reprobada. Entrada: 7 Salida: Aprobado

```python
cal = float(input("Ingresa tu calificación (0-10): "))

if(cal >= 6):
    print("Esta aprobada")
else:
    print("Esta reprobada")
```

**Ejercicio 4: Clasificación por rango de edad con elif**

Enunciado: Crea un programa que clasifique a una persona según su edad: menor de 3 años "Bebé", de 3 a 12 "Niño", de 13 a 17 "Adolescente" y de 18 en adelante "Adulto". Usa elif. Entrada: 15 Salida: Adolescente

```python
age = int(input("Ingresa tu edad: "))

if(age >= 18):
    print("Adulto")
elif(age >= 13):
    print("Adolecentes")
elif(age >= 3):
    print("Niño")
else:
    print("Bebé")

```

**Ejercicio 5: Descuento por tipo de visitante**

Enunciado: Crea un programa que pida el tipo de visitante (adulto_mayor, profesor, estudiante u otro) y calcule el precio de un boleto base de $45 aplicando solo un descuento: adulto mayor 12%, profesor 10%, estudiante 10%, otro 0%. Entrada: estudiante Salida: Precio final: $40.50

```python
tipo_visitante = input("Ingrese que tipo de visitante es: ")
precio_boleto = 45

if(tipo_visitante == "adulto_mayor"):
    precio_final = precio_boleto * 0.88
elif(tipo_visitante == "profesor"):
    precio_final = precio_boleto * 0.90
elif(tipo_visitante == "estudiante"):
    precio_final = precio_boleto * 0.90
else:
    precio_final = precio_boleto

print(f"Total a pagar: {precio_final:.2f} ")
```
